Создать базовый класс CreditCard в C#, который будет представлять информацию
о кредитных картах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    protected decimal balance;

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        balance = initialBalance;
    }

    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber}, Владелец: {HolderName}, Срок действия: {ExpiryDate}";
    }

    public virtual void Pay(decimal amount)
    {
        if (balance >= amount)
        {
            balance -= amount;
            Console.WriteLine($"Успешно оплачено: {amount} руб. Баланс: {balance} руб.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств на карте!");
        }
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Текущий баланс: {balance} руб.");
    }
}

public class GoldCard : CreditCard
{
    public int BonusMiles { get; set; }

    public GoldCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        BonusMiles = bonusMiles;
    }

    public override void Pay(decimal amount)
    {
        base.Pay(amount);
        int earnedMiles = (int)(amount / 100);
        BonusMiles += earnedMiles;
        Console.WriteLine($"Начислено бонусных миль: {earnedMiles}. Всего миль: {BonusMiles}");
    }
}

public class PremiumCard : CreditCard
{
    public string SupportAssistant { get; set; }

    public PremiumCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string supportAssistant)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        SupportAssistant = supportAssistant;
    }

    public override void CheckBalance()
    {
        base.CheckBalance();
        Console.WriteLine($"Персональный ассистент поддержки {SupportAssistant} готов помочь вам с любыми вопросами по балансу.");
    }
}

public class CorporateCard : CreditCard
{
    public string Company { get; set; }

    public CorporateCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string company)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        Company = company;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}";
    }
}

GoldCard gold = new GoldCard("4000 1234 5678 9010", "Иван Иванов", "12/28", 50000m, 150);
Console.WriteLine(gold.GetInfo());
gold.Pay(5000m);
Console.WriteLine();

PremiumCard premium = new PremiumCard("5555 4444 3333 2222", "Анна Смирнова", "05/29", 150000m, "Елена");
Console.WriteLine(premium.GetInfo());
premium.CheckBalance();
Console.WriteLine();

CorporateCard corporate = new CorporateCard("2200 1111 9999 8888", "Петр Петров", "10/27", 300000m, "ООО «ТехноПром»");
Console.WriteLine(corporate.GetInfo());
corporate.Pay(10000m);
```

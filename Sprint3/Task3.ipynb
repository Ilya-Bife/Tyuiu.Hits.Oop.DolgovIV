```
public class CreditCard
{
    private string _cardNumber;
    private string _holderName;
    private string _expiryDate;
    protected decimal balance;

    public string Currency { get; set; } = "руб.";
    public decimal CreditLimit { get; set; }
    public bool IsActive { get; set; } = true;
    public string PinCode { get; set; } = "0000";
    public int RiskLevel { get; set; } = 1;
    public string BankName { get; set; } = "Сбер";
    public bool ContactlessEnabled { get; set; } = true;

    public string CardNumber
    {
        get => _cardNumber;
        set => _cardNumber = string.IsNullOrWhiteSpace(value) ? "Не указан" : value;
    }

    public string HolderName
    {
        get => _holderName;
        set => _holderName = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value;
    }

    public string ExpiryDate
    {
        get => _expiryDate;
        set => _expiryDate = string.IsNullOrWhiteSpace(value) ? "00/00" : value;
    }

    public decimal Balance
    {
        get => balance;
        set
        {
            if (value >= 0)
                balance = value;
            else
                throw new ArgumentOutOfRangeException("Баланс не может быть отрицательным!");
        }
    }

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string currency, decimal creditLimit)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = initialBalance;
        Currency = currency;
        CreditLimit = creditLimit;
    }

    public virtual void ActivateCard() => IsActive = true;
    public virtual void BlockCard() => IsActive = false;
    public virtual void ChangePin(string newPin) => PinCode = newPin;

    public virtual void Pay(decimal amount)
    {
        if (balance >= amount)
        {
            balance -= amount;
            Console.WriteLine($"Успешно оплачено: {amount} {Currency}. Баланс: {balance} {Currency}.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств!");
        }
    }

    public virtual void Pay(decimal amount, string description)
    {
        if (balance >= amount)
        {
            balance -= amount;
            Console.WriteLine($"Успешно оплачено ({description}): {amount} {Currency}. Баланс: {balance} {Currency}.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств для платежа с описанием!");
        }
    }

    public virtual void Pay(string category, decimal amount)
    {
        if (balance >= amount)
        {
            balance -= amount;
            Console.WriteLine($"Успешно оплачена категория [{category}] на сумму {amount} {Currency}. Баланс: {balance} {Currency}.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств в категории!");
        }
    }

    public virtual void ProcessTransaction()
    {
        Console.WriteLine($"Карта: {CardNumber}, Владелец: {HolderName}, Срок действия: {ExpiryDate}");
    }
}

public class GoldCard : CreditCard
{
    private int _bonusMiles;

    public decimal CashbackPercentage { get; set; } = 1.5m;
    public int LoungeAccessCount { get; set; } = 4;
    public string PersonalManager { get; set; } = "Елена";
    public bool PrioritySupport { get; set; } = true;

    public int BonusMiles
    {
        get => _bonusMiles;
        set => _bonusMiles = value >= 0 ? value : 0;
    }

    public GoldCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string currency, decimal creditLimit, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, initialBalance, currency, creditLimit)
    {
        BonusMiles = bonusMiles;
    }

    public override void Pay(decimal amount)
    {
        base.Pay(amount);
        int earnedMiles = (int)(amount / 100);
        BonusMiles += earnedMiles;
        Console.WriteLine($"Начислено бонусных миль: {earnedMiles}. Всего миль: {BonusMiles}");
    }

    public override void Pay(decimal amount, string description)
    {
        base.Pay(amount, description);
        int earnedMiles = (int)(amount / 100);
        BonusMiles += earnedMiles;
        Console.WriteLine($"Начислено бонусных миль по описанию: {earnedMiles}. Всего миль: {BonusMiles}");
    }

    public override void Pay(string category, decimal amount)
    {
        base.Pay(category, amount);
        int earnedMiles = (int)(amount / 50);
        BonusMiles += earnedMiles;
        Console.WriteLine($"Начислено двойных миль за категорию [{category}]: {earnedMiles}. Всего миль: {BonusMiles}");
    }

    public override void ProcessTransaction()
    {
        Console.WriteLine($"Карта: {CardNumber}, Владелец: {HolderName}, Срок действия: {ExpiryDate}");
        Console.WriteLine($"Текущий баланс: {balance} {Currency}.");
        Console.WriteLine($"Персональный ассистент поддержки {PersonalManager} готов помочь вам с любыми вопросами по балансу.");
    }
}

public class CardCollection<T> where T : CreditCard
{
    private List<T> _cards = new List<T>();

    public void AddCard(T card)
    {
        _cards.Add(card);
    }

    public void ProcessAllTransactions()
    {
        foreach (var card in _cards)
        {
            card.ProcessTransaction();
            Console.WriteLine();
        }
    }
}

CardCollection<CreditCard> collection = new CardCollection<CreditCard>();
collection.AddCard(new CreditCard("4000 1234 5678 9010", "Иван Иванов", "12/28", 50000m, "руб.", 10000m));
collection.AddCard(new GoldCard("5555 4444 3333 2222", "Анна Смирнова", "05/29", 150000m, "руб.", 50000m, 150));

collection.ProcessAllTransactions();

CreditCard testCard = new CreditCard("4000 1234 5678 9010", "Иван Иванов", "12/28", 50000m, "руб.", 10000m);
testCard.Pay(5000m);
```

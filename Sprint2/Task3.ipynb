демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;

public interface ICashback
{
    decimal CashbackBalance { get; set; }
    void ApplyCashback();
    void WithdrawCashback();
    decimal CalculateMonthlyCashback(decimal spent);
}

public interface IInsured
{
    decimal InsuranceSum { get; set; }
    void TriggerInsuranceEvent();
    void RenewInsurance();
    string GetInsuranceStatus();
}

public class CreditCard
{
    private string _cardNumber;
    private string _holderName;
    private string _expiryDate;
    protected decimal balance;

    public string Currency { get; set; } = "RUB";
    public decimal CreditLimit { get; set; }
    public bool IsActive { get; set; } = true;
    public string PinCode { get; set; } = "0000";

    public string CardNumber
    {
        get => _cardNumber;
        set => _cardNumber = string.IsNullOrWhiteSpace(value) ? "Не указан" : value;
    }

    public string HolderName
    {
        get => _holderName;
        set => _holderName = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value;
    }

    public string ExpiryDate
    {
        get => _expiryDate;
        set => _expiryDate = string.IsNullOrWhiteSpace(value) ? "00/00" : value;
    }

    public decimal Balance
    {
        get => balance;
        set
        {
            if (value >= 0)
                balance = value;
            else
                throw new ArgumentOutOfRangeException("Баланс не может быть отрицательным!");
        }
    }

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string currency, decimal creditLimit)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = initialBalance;
        Currency = currency;
        CreditLimit = creditLimit;
    }

    public virtual void ActivateCard() => IsActive = true;
    public virtual void BlockCard() => IsActive = false;
    public virtual void ChangePin(string newPin) => PinCode = newPin;

    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber}, Владелец: {HolderName}, Валюта: {Currency}, Активна: {IsActive}";
    }

    public virtual void Pay(decimal amount)
    {
        if (balance >= amount)
        {
            balance -= amount;
            Console.WriteLine($"Успешно оплачено: {amount} {Currency}. Баланс: {balance} {Currency}.");
        }
        else
        {
            Console.WriteLine("Недостаточно средств на карте!");
        }
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Текущий баланс: {balance} {Currency}.");
    }
}

public class GoldCard : CreditCard
{
    private int _bonusMiles;

    public decimal CashbackPercentage { get; set; } = 1.5m;
    public int LoungeAccessCount { get; set; } = 4;
    public string PersonalManager { get; set; } = "Стандарт";
    public bool PrioritySupport { get; set; } = true;

    public int BonusMiles
    {
        get => _bonusMiles;
        set => _bonusMiles = value >= 0 ? value : 0;
    }

    public GoldCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string currency, decimal creditLimit, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, initialBalance, currency, creditLimit)
    {
        BonusMiles = bonusMiles;
    }

    public void UseLounge()
    {
        LoungeAccessCount = Math.Max(0, LoungeAccessCount - 1);
        Console.WriteLine($"Доступ в бизнес-зал использован. Осталось посещений: {LoungeAccessCount}");
    }

    public void UpgradeManager(string managerName)
    {
        PersonalManager = managerName;
        Console.WriteLine($"Персональный менеджер обновлен на: {managerName}");
    }

    public void ConvertMilesToCash(int miles)
    {
        if (BonusMiles >= miles)
        {
            BonusMiles -= miles;
            balance += miles * 0.5m;
            Console.WriteLine($"Конвертировано {miles} миль в деньги.");
        }
    }

    public override void Pay(decimal amount)
    {
        base.Pay(amount);
        int earnedMiles = (int)(amount / 100);
        BonusMiles += earnedMiles;
        Console.WriteLine($"Начислено бонусных миль: {earnedMiles}. Всего миль: {BonusMiles}");
    }
}

public class VipGoldCard : GoldCard
{
    public string SafeBoxNumber { get; set; } = "A-12";
    public decimal InvestmentYieldRate { get; set; } = 8.5m;
    public bool Concierge247 { get; set; } = true;
    public int FreeAirportTransfers { get; set; } = 2;

    public VipGoldCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string currency, decimal creditLimit, int bonusMiles)
        : base(cardNumber, holderName, expiryDate, initialBalance, currency, creditLimit, bonusMiles)
    {
    }

    public void BookTransfer()
    {
        if (FreeAirportTransfers > 0)
        {
            FreeAirportTransfers--;
            Console.WriteLine("Трансфер в аэропорт успешно забронирован.");
        }
    }

    public void OpenSafeBox()
    {
        Console.WriteLine($"Банковская сейфовая ячейка №{SafeBoxNumber} открыта для владельца {HolderName}.");
    }

    public void ConsultInvestmentAdvisor()
    {
        Console.WriteLine($"Инвестиционный советник готов обсудить доходность {InvestmentYieldRate}%.");
    }

    public override void Pay(decimal amount)
    {
        base.Pay(amount);
        Console.WriteLine("[VIP привилегия]: Сработал повышенный уровень кешбэка.");
    }
}

public class CorporateCard : CreditCard, ICashback, IInsured
{
    public string Company { get; set; }

    public string TaxNumber { get; set; } = "1234567890";
    public int EmployeeLimit { get; set; } = 10;
    public int BillingCycleDays { get; set; } = 30;
    public decimal CashbackBalance { get; set; } = 0;
    public decimal InsuranceSum { get; set; } = 1000000m;

    public CorporateCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string currency, decimal creditLimit, string company)
        : base(cardNumber, holderName, expiryDate, initialBalance, currency, creditLimit)
    {
        Company = company;
    }

    public void ApplyCashback()
    {
        Balance += CashbackBalance;
        Console.WriteLine($"Кэшбэк {CashbackBalance} зачислен на корпоративный баланс.");
        CashbackBalance = 0;
    }

    public void WithdrawCashback() => Console.WriteLine("Кэшбэк выведен на расчетный счет компании.");
    public decimal CalculateMonthlyCashback(decimal spent) => spent * 0.02m;

    public void TriggerInsuranceEvent() => Console.WriteLine($"Зарегистрирован страховой случай на сумму до {InsuranceSum} руб.");
    public void RenewInsurance() => Console.WriteLine("Корпоративная страховка успешно продлена.");
    string IInsured.GetInsuranceStatus() => "Действительна";

    public void ApproveExpense(decimal amount) => Console.WriteLine($"Расход на сумму {amount} согласован для компании {Company}.");
    public void GenerateTaxReport() => Console.WriteLine($"Налоговый отчет компании {Company} сгенерирован.");
    public void AuditTransactions() => Console.WriteLine("Аудит транзакций успешно завершен без нарушений.");

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, Лимит сотрудников: {EmployeeLimit}";
    }
}

VipGoldCard vipCard = new VipGoldCard("1111 2222 3333 4444", "Александр Смирнов", "12/30", 500000m, "USD", 10000m, 500);
CorporateCard corpCard = new CorporateCard("9999 8888 7777 6666", "Ольга Васильева", "08/28", 1200000m, "RUB", 500000m, "ООО «Альфа»");

Console.WriteLine("=== Информация по VIP-карте (Сложное наследование) ===");
Console.WriteLine(vipCard.GetInfo());
vipCard.OpenSafeBox();
vipCard.BookTransfer();
vipCard.Pay(200);

Console.WriteLine("\n=== Информация по Корпоративной карте (Множественное наследование) ===");
Console.WriteLine(corpCard.GetInfo());
corpCard.ApproveExpense(45000m);
corpCard.TriggerInsuranceEvent();
corpCard.AuditTransactions();
```
